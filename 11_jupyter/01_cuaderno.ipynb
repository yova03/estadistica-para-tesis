# Cuaderno de ejemplo (Jupyter)

Un cuaderno tiene **celdas**: unas de texto (como esta) y otras de codigo.
Cada celda de codigo se ejecuta por separado y el resultado aparece debajo.

**Como se abre:**

2. Desde esta carpeta ejecuta: `jupyter notebook`
3. Se abre el navegador; haz clic en `01_cuaderno.ipynb`

```python
# Celda 1: leer los datos y espiar la tabla
import pandas as pd

datos = pd.read_csv("../datos/mediciones.csv")
datos.head()
```

## Estadistica rapida

Ahora calculamos el promedio por sitio y hacemos una prueba t de Student,
igual que en los ejemplos de R y de Python.

```python
# Celda 2: promedios por sitio
datos.groupby("sitio")["peso_g"].agg(["count", "mean", "std"]).round(2)
```

```python
# Celda 3: prueba t de Student
from scipy import stats

grupo_a = datos.loc[datos["sitio"] == "Sitio A", "peso_g"]
grupo_b = datos.loc[datos["sitio"] == "Sitio B", "peso_g"]

t, p = stats.ttest_ind(grupo_a, grupo_b)
print(f"t = {t:.3f}")
print(f"p = {p:.6f}")
```

## Grafico

Y por ultimo, un grafico que queda guardado en la memoria del cuaderno.
Este es el uso tipico de Jupyter en investigacion: explorar y analizar.

```python
# Celda 4: grafico de cajas
%matplotlib inline
import matplotlib.pyplot as plt

datos.boxplot(column="peso_g", by="sitio")
plt.title("Peso por sitio")
plt.suptitle("")
plt.xlabel("Sitio")
plt.ylabel("Peso (g)")
plt.show()
```
